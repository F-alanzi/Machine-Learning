
## 3. Handling Missing Values

### 3.1 Detect Missing Values


```python
df.isna().sum().sort_values(ascending=False).head(20) # Count missing values and show top 20
```


Columns like **PoolQC, MiscFeature, Alley, Fence** have many missing values.

We must decide whether to:
- Drop the column
- Drop rows
- Impute values


### Strategy: Median Imputation for SalePrice-related numeric features

```python
df_imputed = df.copy() # Create a copy of the dataframe
numeric_cols = df_imputed.select_dtypes(include=np.number).columns # Identify numeric columns
for col in numeric_cols:
    df_imputed[col] = df_imputed[col].fillna(df_imputed[col].median()) # Fill numeric NAs with median
object_cols = df_imputed.select_dtypes(include='object').columns # Identify categorical columns
for col in object_cols:
    df_imputed[col] = df_imputed[col].fillna("None") # Fill categorical NAs with 'None'
print(f"Remaining missing values: {df_imputed.isna().sum().sum()}") # Check total missing count
display(df_imputed.head()) # Show the cleaned data
```


## 4. Handling Outliers

We detect outliers in **SalePrice** using boxplot and IQR method.


```python
plt.figure(figsize=(6,4)) # Initialize a figure with specific dimensions
sns.boxplot(x=df['SalePrice']) # Create a boxplot for SalePrice
plt.title("Boxplot of SalePrice") # Add a title to the boxplot
plt.show() # Render the plot
```

```python
Q1 = df['SalePrice'].quantile(0.25) # Calculate the 25th percentile
Q3 = df['SalePrice'].quantile(0.75) # Calculate the 75th percentile
IQR = Q3 - Q1 # Calculate Interquartile Range
lower = Q1 - 1.5 * IQR # Define lower outlier bound
upper = Q3 + 1.5 * IQR # Define upper outlier bound
df_no_outliers = df[(df['SalePrice'] >= lower) & (df['SalePrice'] <= upper)] # Filter out outliers
print("Original shape:", df.shape) # Print count before cleaning
print("After removing outliers:", df_no_outliers.shape) # Print count after cleaning
```


## 5. Data Transformation – Normalization

We apply Min-Max and Z-score normalization to:
- SalePrice
- GrLivArea


```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler # Import scaling tools
features = df[['SalePrice', 'GrLivArea']].copy() # Select features to scale
minmax = MinMaxScaler() # Initialize Min-Max scaler (0 to 1)
df_minmax = minmax.fit_transform(features) # Apply Min-Max scaling
standard = StandardScaler() # Initialize Z-score scaler (mean=0, std=1)
df_standard = standard.fit_transform(features) # Apply Z-score scaling
```


## Correlation Heatmap (Before PCA)


```python
plt.figure(figsize=(6,4)) # Set the figure size for the heatmap
sns.heatmap(df[['SalePrice','GrLivArea','LotArea','OverallQual']].corr(), annot=True, cmap="coolwarm") # Plot correlation matrix with values
plt.title("Correlation Heatmap") # Set the title of the heatmap
plt.show() # Show the final heatmap
```


## 6. Data Reduction – Principal Component Analysis (PCA)


```python
from sklearn.decomposition import PCA # Import PCA for dimensionality reduction
X = df[['SalePrice','GrLivArea','LotArea','OverallQual']].dropna() # Prepare features and drop NAs
scaler = StandardScaler() # Initialize scaler for PCA
X_scaled = scaler.fit_transform(X) # Scale data before applying PCA
pca = PCA(n_components=2) # Initialize PCA for 2 components
principal_components = pca.fit_transform(X_scaled) # Transform data into principal components
print("Explained Variance Ratio:", pca.explained_variance_ratio_) # Show variance captured by each component
```

```python
plt.figure(figsize=(6,4)) # Set the figure dimensions
plt.scatter(principal_components[:,0], principal_components[:,1]) # Create scatter plot of the two PCA components
plt.title("PCA Projection") # Set the title of the scatter plot
plt.xlabel("Principal Component 1") # Label the x-axis
plt.ylabel("Principal Component 2") # Label the y-axis
plt.show() # Display the scatter plot
```
